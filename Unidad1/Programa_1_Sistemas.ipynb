```python
# Ejemplo de programa con 2 condiciones, cuando la primera no se cumple,
# Se vuelve a revisar otra condicion
variable1 = 70
variable2 = 70
if variable1 > variable2:
    print( variable1, "es mayor que", variable2 )
elif variable1 == variable2:
    print(variable1, "es igual que", variable2)
else:
    print(variable2, "es mayor que", variable1)
```

# **Reglas prácticas para evitar errores de lógica:**

- Ordenar condiciones de menor a mayor (o viceversa), pero siempre de forma consistente.
- Usar elif en lugar de varios if independientes, para que solo una condición se ejecute.
- Dejar el else para el caso “por defecto” (cuando ninguna condición anterior se cumple).
- Pensar en los rangos como intervalos que no se solapan.

```python
tabla = int(input("Que tabla de multiplicar quieres"))
inicio = int(input("Dame el limite inicial"))
final = int(input("Dame el limite final"))
incrementos = int(input("Dame el numero de los incrementos"))
# Agregamos la instruccion while para que valide que el valor
# inicla seas mayor que el final, de lo contrario me pedira
# indefinidamente el valor final hasta que sea mayor
while inicio >= final:
  print("El limite final debe ser mayor que el inicial")
  final = int(input("Dame el limite final"))
for i in range(inicio, final, incrementos):
  print(tabla, "*", i, "=", tabla*i)
```

# **Listas en Python**

```python
# Podemos acceder a cada elemento o item de la lista a traves
# del indice o su posicion. Las listas siempre empiezan en la posicion cero
alumnos = ["Ana", "Luis", "Carlos"]
print(alumnos)
print(alumnos[0])
print(alumnos[2])
```

```python
# pdemos modificar el valor de un elemento de la lista
# Usando el nombre de la lista y el indice y le asignamos
# el nuevo valor
print(alumnos)
alumnos[1] = "Juan"
print(alumnos)
```

```python
# Podemos especificar indices negativos y cuando se ponene valores
# negativos se lee ose accede a la lista desde el final
# empezando con -1
print(alumnos)
print(alumnos[-1])
print(alumnos[-2])
```

```python
# El metodo append agrega un elemento al final de la lista
print(alumnos)
alumnos.append("Sandra")
print(alumnos)
```

```python
print(alumnos)
alumnos.insert(2, "Sofía")
print(alumnos)
```

```python
# remove borra un elemento de la lista especificando el
# valor, si existen valores repetidos, borra el primer valor
# que encuentra
print(alumnos)
alumnos.remove("Sofía")
print(alumnos)
```

```python
print(alumnos[2])
# la siguiente linea marca error porque queremos acceder a
# un indice que no existe. la lista tiene unicamente 6 elementos y
# queremos acceder al elemneto 7 que no existe
print(alumnos[7])
```

```python
# pop me permite borrra un elemento de la lista
# especificando su posicion
print(alumnos)
alumnos.pop(2)
print(alumnos)
```

```python
# del me permite borrar un elemento de la lista
# especificando su posicion
print(alumnos)
del alumnos[3]
print(alumnos)
```

```python
# la funcion len me permite conocer el numero de
# elementos de la lista
print("Número de elementos en la lista", len(alumnos) )
```

```python
# Ciclo para recorrer la lista y hacer algun proceso
alumnos.append("Sandra")
alumnos.append("Carlos")
print(alumnos)
for x in alumnos:
  print(x)
  if x == "Sofía":
    print("Encontramos a Sofia")
    break   # break forza a salir de un ciclo
```
