# **Reglas prácticas** para **evitar errores** de lógica en un **if** al establecer las condiciones

- Ordenar condiciones de menor a mayor (o viceversa), pero siempre de forma consistente.
- Usar **elif en lugar de varios if independientes**, para que solo una condición se ejecute.
- Dejar el **else** para el caso “por defecto” (cuando ninguna condición anterior se cumple).
- Pensar en los rangos como **intervalos que no se solapan**.

```python
##   Instruccion for sirve para repetir instrucciones
tabla = 7
for i in range(10):
  print(tabla, "*", i, "=", tabla*i)
```

# **Tema de listas**

```python
alumnos = ["Ana", "Luis", "Carlos", "Juan"]
print(alumnos)
print(alumnos[0])
print(alumnos[3])
```

```python
print(alumnos[5]) # esta instruccion marca error porque el indice o posicion no existe
```

```python
# Modificar un valor de la lista en un indice especifico
print(alumnos)
alumnos[1]= " María Luisa"
print(alumnos)
```

```python
# El metodo append agrega un elemento al final de la lista
print(alumnos)
alumnos.append("Karla")
print(alumnos)
```

```python
# El metodo insert agrega un elemento a la lista en la posicion especificada
print(alumnos)
alumnos.insert(2, "Cristina")
print(alumnos)
```

```python
# remove elimina un elemento especificando el valor del elemento
# debe coincidir exactamente mayusculas y minusculas
print(alumnos)
alumnos.remove("Ana")
print(alumnos)
```

```python
# esta marcaría error porque no existe el valor "ana" en la lista
print(alumnos)
alumnos.remove("ana")
print(alumnos)
```

```python
# pop elimina el elemento o item en la posicion especificada
print(alumnos)
alumnos.pop(2)
print(alumnos)
```

```python
# Metodo del tambien elimina el elemento o item en la posicion especificada
print(alumnos)
del alumnos[3]
print(alumnos)
```

```python
# la funcion len me permite comnocer cvuangtos elementos tiene una lista
print("Numero de Elementos en la lista de alumnos", len(alumnos))
```

```python
print(alumnos)
alumnos.append("Ana")
alumnos.append("Ernesto")
# Recorrer la lista de alumnos uno por uno
print("\n Imprimir la lista de alumnos 1 por 1")
for x in alumnos:
  print(x)
```
